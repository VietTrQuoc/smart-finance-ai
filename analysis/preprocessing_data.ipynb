# Ánh xạ đặc trưng dữ liệu giao dịch để phân cụm khách hàng

Tài liệu này tổng hợp các cột dữ liệu có thể dùng để tạo đặc trưng ở mức `customer_id`, phục vụ bài toán phân cụm khách hàng và khuyến nghị sản phẩm tài chính như thẻ hoàn tiền, thẻ du lịch, bảo hiểm du lịch, BNPL/trả góp, vay tiêu dùng, bảo hiểm thiết bị và voucher/cashback thương mại.

## 1. `amount`

### Các feature nên tạo

| Feature | Ý nghĩa |
|---|---|
| `total_amount` | Tổng chi tiêu |
| `mean_amount` | Giao dịch trung bình |
| `median_amount` | Mức chi tiêu điển hình |
| `max_amount` | Giao dịch lớn nhất |
| `std_amount` | Biến động chi tiêu |
| `high_value_txn_ratio` | Tỷ lệ giao dịch lớn |

### Lý do nên ánh xạ

Cột `amount` phản ánh trực tiếp **quy mô chi tiêu** và **sức mua** của khách hàng. Đây là nhóm đặc trưng rất quan trọng để phân biệt khách hàng chi tiêu thấp, trung bình, cao hoặc có nhiều giao dịch giá trị lớn.

---

## 2. `merchant_category`

### Cách ánh xạ

Tạo ma trận:

```text
customer_id × merchant_category
```

Mỗi dòng là một khách hàng, mỗi cột là một nhóm merchant, giá trị trong ô là số lượng hoặc tỷ lệ giao dịch của khách hàng đó trong nhóm merchant tương ứng.

Ví dụ:
| customer_id | Retail | Grocery | Travel | Electronics | Entertainment |
|---|---:|---:|---:|---:|---:|
| C001 | 20 | 15 | 3 | 1 | 6 |
| C002 | 2 | 4 | 18 | 0 | 2 |
| C003 | 10 | 5 | 0 | 8 | 3 |

### Lý do nên ánh xạ

`merchant_category` là cột quan trọng nhất để nhận diện **mục đích chi tiêu** của khách hàng. Nó cho biết khách hàng thường chi tiền vào đâu: mua sắm, ăn uống, du lịch, giải trí, điện tử, nhu yếu phẩm, v.v.

---

## 3. `channel`

### Các feature nên tạo

| Feature | Ý nghĩa |
|---|---|
| `mobile_ratio` | Tỷ lệ mobile |
| `web_ratio` | Tỷ lệ web |
| `pos_ratio` | Tỷ lệ POS |
| `online_ratio = mobile + web` | Mức độ số hóa |

### Lý do nên ánh xạ

`channel` phản ánh **kênh giao dịch chính** của khách hàng. Đây là tín hiệu tốt để phân biệt khách hàng thiên về giao dịch online, mobile, hay giao dịch trực tiếp tại điểm bán.

---

## 4. `card_type`

### Các feature nên tạo

| Feature | Ý nghĩa |
|---|---|
| `basic_card_ratio` | Tỷ lệ giao dịch bằng Basic Credit |
| `gold_card_ratio` | Tỷ lệ giao dịch bằng Gold Credit |
| `premium_card_ratio` | Nếu có Platinum/Premium |
| `unique_card_type_count` | Số loại thẻ từng dùng |

### Lý do nên ánh xạ

`card_type` cho biết khách hàng đang sử dụng loại thẻ nào. Đây là tín hiệu hữu ích để xác định khả năng **upsell**, **cross-sell**, hoặc gợi ý sản phẩm phù hợp với phân khúc hiện tại của khách hàng.

---

## 5. `country`, `city`, `currency`

### Các feature nên tạo

| Feature | Ý nghĩa |
|---|---|

| `unique_cities` | Số thành phố giao dịch |
| `unique_currencies` | Số loại tiền tệ |
| `foreign_txn_ratio` | Tỷ lệ giao dịch nước ngoài |
| `top_country_ratio` | Mức tập trung ở 1 quốc gia |
| `city_diversity` | Đa dạng thành phố |

### Lý do nên ánh xạ

Nhóm cột `country`, `city`, `currency` phản ánh **hành vi địa lý** và **mức độ giao dịch quốc tế** của khách hàng. Đây là nhóm đặc trưng rất quan trọng cho các sản phẩm liên quan đến du lịch, chi tiêu quốc tế và bảo hiểm.

## 6. `distance_from_home`

### Các feature nên tạo

| Feature | Ý nghĩa |
|---|---|
| `distance_from_home_ratio` | Tỷ lệ giao dịch xa nhà |

### Lý do nên ánh xạ

`distance_from_home` là tín hiệu trực tiếp cho biết khách hàng có thường xuyên giao dịch ngoài khu vực quen thuộc hoặc ngoài quốc gia cư trú hay không. Đây là một trong những feature mạnh nhất cho các sản phẩm liên quan đến du lịch và bảo vệ rủi ro khi đi xa.

## 9. Kết luận

Các cột trên đều có giá trị phân cụm vì chúng mô tả các chiều hành vi khác nhau của khách hàng:

```text
amount              → quy mô chi tiêu
merchant_category   → mục đích chi tiêu
channel             → kênh giao dịch
card_type           → loại thẻ/phân khúc tài chính
country/city/currency → hành vi địa lý/quốc tế
distance_from_home  → mức độ di chuyển/xa nhà
```

Khi kết hợp lại, bạn có thể phân cụm khách hàng thành các nhóm có ý nghĩa như:

```text
Shopping users
Travel users
Digital-first users
Big-ticket buyers
High spenders
Low activity users
```

Sau đó dùng từng cụm để tạo danh sách sản phẩm ứng viên và tính điểm Top K cho từng khách hàng.


````python
# ----------------------------------------------------------
# 0. LOẠI BỎ KHÁCH HÀNG CÓ >50% GIAO DỊCH FRAUD
# ----------------------------------------------------------
fraud_rate = df.groupby("customer_id")["is_fraud"].mean()
bad_customers = fraud_rate[fraud_rate > 0.5].index
n_before = df["customer_id"].nunique()
df = df[~df["customer_id"].isin(bad_customers)]
n_after = df["customer_id"].nunique()
print(f"🔍 Lọc fraud: {n_before:,} → {n_after:,} khách (đã loại {n_before - n_after:,} khách có >50% fraud)")

# ----------------------------------------------------------
# 1. AMOUNT — Đặc trưng chi tiêu
# ----------------------------------------------------------
amount_agg = df.groupby("customer_id")["amount"].agg(
    total_amount="sum",
    mean_amount="mean",
    median_amount="median",
    max_amount="max",
    std_amount="std",
).reset_index()

# Tỷ lệ giao dịch có giá trị "lớn" (>= ngưỡng 75th percentile toàn tập)
high_value_cutoff = df["amount"].quantile(0.75)
amount_agg["high_value_txn_ratio"] = df.groupby("customer_id")["amount"].apply(
    lambda s: (s >= high_value_cutoff).mean()
).values

# ----------------------------------------------------------
# 2. MERCHANT_CATEGORY — Pivot customer × merchant_category (tỷ lệ)
# ----------------------------------------------------------
merchant_pivot = (
    df.pivot_table(
        index="customer_id",
        columns="merchant_category",
        aggfunc="size",
        fill_value=0,
    )
    .pipe(lambda t: t.div(t.sum(axis=1), axis=0))  # chuẩn hóa thành tỷ lệ
    .add_prefix("merchant_")
    .reset_index()
)
merchant_pivot.columns.name = None  # xoá tên cột index

# ----------------------------------------------------------
# 3. CHANNEL — Tỷ lệ kênh giao dịch
# ----------------------------------------------------------
channel_df = (
    df.groupby("customer_id")["channel"]
    .value_counts(normalize=True)
    .unstack(fill_value=0)
    .rename(columns={"mobile": "mobile_ratio", "web": "web_ratio", "pos": "pos_ratio"})
    .reset_index()
)
channel_df["online_ratio"] = channel_df["mobile_ratio"] + channel_df["web_ratio"]

# ----------------------------------------------------------
# 4. CARD_TYPE — Tỷ lệ loại thẻ + số loại thẻ từng dùng
# ----------------------------------------------------------
# Gom nhóm các loại thẻ về 3 nhóm chính
CARD_GROUP_MAP = {
    "Basic Credit":   "basic",
    "Basic Debit":    "basic",
    "Gold Credit":    "gold",
    "Premium Debit":  "premium",
    "Platinum Credit": "premium",
}
card_group = df["card_type"].map(CARD_GROUP_MAP)

card_ratio = (
    card_group.groupby(df["customer_id"])
    .value_counts(normalize=True)
    .unstack(fill_value=0)
    .rename(columns={"basic": "basic_card_ratio", "gold": "gold_card_ratio", "premium": "premium_card_ratio"})
    .reset_index()
)

unique_card = (
    df.groupby("customer_id")["card_type"]
    .nunique()
    .reset_index(name="unique_card_type_count")
)

# ----------------------------------------------------------
# 5. COUNTRY, CITY, CURRENCY — Đa dạng địa lý
# ----------------------------------------------------------
geo_agg = df.groupby("customer_id").agg(
    unique_countries=("country", "nunique"),
    unique_cities=("city", "nunique"),
    unique_currencies=("currency", "nunique"),
    txn_count=("transaction_id", "count"),
).reset_index()

# Tỷ lệ giao dịch nước ngoài (distance_from_home = 1)
foreign_ratio = (
    df.groupby("customer_id")["distance_from_home"]
    .mean()
    .reset_index(name="foreign_txn_ratio")
)

# Mức độ tập trung vào 1 quốc gia (tỷ lệ giao dịch ở quốc gia xuất hiện nhiều nhất)
top_country = (
    df.groupby("customer_id")["country"]
    .apply(lambda s: s.value_counts(normalize=True).iloc[0])
    .reset_index(name="top_country_ratio")
)

# Đa dạng thành phố = unique_cities / total_txn (càng gần 1 càng đa dạng)
geo_agg["city_diversity"] = geo_agg["unique_cities"] / geo_agg["txn_count"]
geo_agg = geo_agg.drop(columns="txn_count")  # không cần giữ cột phụ

# ----------------------------------------------------------
# 6. DISTANCE_FROM_HOME — Giao dịch xa nhà
# ----------------------------------------------------------
dist_agg = df.groupby("customer_id").agg(
    distance_from_home_ratio=("distance_from_home", "mean"),
    distance_from_home_count=("distance_from_home", "sum"),
).reset_index()

# Tỷ lệ xa nhà trong 30 ngày gần nhất (time‑window)
df["_ts"] = pd.to_datetime(df["timestamp"], format="mixed")
last_ts = df["_ts"].max()
mask_recent = df["_ts"] >= (last_ts - pd.Timedelta(days=30))
recent_dist = (
    df[mask_recent]
    .groupby("customer_id")["distance_from_home"]
    .mean()
    .reset_index(name="recent_distance_from_home_ratio")
)

# ----------------------------------------------------------
# 7. GHÉP TOÀN BỘ ĐẶC TRƯNG VỀ 1 BẢNG customer-level
# ----------------------------------------------------------
customer_features = (
    amount_agg
    .merge(channel_df, on="customer_id", how="left")
    .merge(card_ratio, on="customer_id", how="left")
    .merge(unique_card, on="customer_id", how="left")
    .merge(geo_agg, on="customer_id", how="left")
    .merge(foreign_ratio, on="customer_id", how="left")
    .merge(top_country, on="customer_id", how="left")
    .merge(dist_agg, on="customer_id", how="left")
    .merge(recent_dist, on="customer_id", how="left")
    .merge(merchant_pivot, on="customer_id", how="left")
)

# Điền NaN = 0 cho các cột tỷ lệ (khách chưa từng dùng kênh/loại thẻ nào đó)
customer_features = customer_features.fillna(0)

# Xoá cột tạm _ts khỏi df gốc để giữ sạch
df = df.drop(columns=["_ts"])

# ----------------------------------------------------------
# KẾT QUẢ
# ----------------------------------------------------------
print("✅ Kích thước bảng customer-level:", customer_features.shape)
print(f"   Số khách hàng: {customer_features['customer_id'].nunique():,}")
print(f"   Số đặc trưng (cột): {customer_features.shape[1]}")
display(customer_features.head(5))
````
