```python
def des(column_name: str) -> str:
    """
    Hàm hiển thị mô tả bằng tiếng Việt của các cột trong bảng dữ liệu transaction.
    """
    # Chuẩn hóa tên cột về dạng chữ thường và xóa khoảng trắng thừa
    col = column_name.strip().lower()
    
    # Từ điển chứa toàn bộ metadata đã được dịch sang tiếng Việt
    schema_descriptions = {
        "transaction_id": "Mã định danh duy nhất cho mỗi giao dịch.",
        "customer_id": "Mã định danh duy nhất cho mỗi khách hàng trong tập dữ liệu.",
        "card_number": "Số thẻ đã được mã hóa/ẩn một phần để bảo mật, liên kết với giao dịch.",
        "timestamp": "Ngày và giờ diễn ra giao dịch.",
        "merchant_category": "Danh mục chung của người bán/đơn vị chấp nhận thẻ (ví dụ: Bán lẻ, Tạp hóa, Du lịch).",
        "merchant_type": "Loại hình cụ thể trong danh mục của người bán (ví dụ: 'online' đối với ngành Bán lẻ).",
        "merchant": "Tên của người bán/đơn vị nơi giao dịch diễn ra.",
        "amount": "Số tiền giao dịch (loại tiền tệ phụ thuộc vào từng quốc gia).",
        "currency": "Loại tiền tệ được sử dụng cho giao dịch (ví dụ: USD, EUR, JPY, VND).",
        "country": "Quốc gia nơi giao dịch diễn ra.",
        "city": "Thành phố nơi giao dịch diễn ra.",
        "city_size": "Quy mô của thành phố (ví dụ: trung bình, lớn).",
        "card_type": "Loại thẻ được sử dụng (ví dụ: Thẻ tín dụng cơ bản, Thẻ tín dụng hạng Vàng).",
        "card_present": "Chỉ báo cho biết thẻ có xuất hiện vật lý trong quá trình giao dịch hay không (thường dùng trong các giao dịch trực tiếp tại máy POS).",
        "device": "Thiết bị được sử dụng để giao dịch (ví dụ: Trình duyệt Chrome, Ứng dụng iOS, Thanh toán qua NFC).",
        "channel": "Kênh thực hiện giao dịch (web, mobile, POS).",
        "device_fingerprint": "Mã định danh duy nhất (dấu vân tay thiết bị) của thiết bị được sử dụng trong giao dịch.",
        "ip_address": "Địa chỉ IP liên quan đến giao dịch.",
        "distance_from_home": "Chỉ báo nhị phân cho biết giao dịch có diễn ra bên ngoài quốc gia cư trú của khách hàng hay không.",
        "high_risk_merchant": "Chỉ báo cho biết danh mục người bán đó có nguy cơ rủi ro gian lận cao hay không (ví dụ: Du lịch, Giải trí).",
        "transaction_hour": "Giờ trong ngày khi giao dịch được thực hiện.",
        "weekend_transaction": "Giá trị Boolean (True/False) cho biết giao dịch có diễn ra vào cuối tuần hay không.",
        
        # Cột dạng từ điển (Dictionary)
        "velocity_last_hour": (
            "Từ điển chứa các chỉ số về tần suất/tốc độ giao dịch trong giờ qua, bao gồm:\n"
            "  - num_transactions: Số lượng giao dịch trong giờ qua của khách hàng này.\n"
            "  - total_amount: Tổng số tiền đã chi tiêu trong giờ qua.\n"
            "  - unique_merchants: Số lượng người bán duy nhất trong giờ qua.\n"
            "  - unique_countries: Số lượng quốc gia duy nhất trong giờ qua.\n"
            "  - max_single_amount: Số tiền lớn nhất của một giao dịch đơn lẻ trong giờ qua."
        ),
        
        # Các thuộc tính con trong velocity_last_hour phòng trường hợp bạn gọi riêng lẻ
        "num_transactions": "[Thuộc velocity_last_hour] Số lượng giao dịch trong giờ qua của khách hàng này.",
        "total_amount": "[Thuộc velocity_last_hour] Tổng số tiền đã chi tiêu trong giờ qua.",
        "unique_merchants": "[Thuộc velocity_last_hour] Số lượng người bán duy nhất trong giờ qua.",
        "unique_countries": "[Thuộc velocity_last_hour] Số lượng quốc gia duy nhất trong giờ qua.",
        "max_single_amount": "[Thuộc velocity_last_hour] Số tiền lớn nhất của một giao dịch đơn lẻ trong giờ qua.",
        
        "is_fraud": "Chỉ báo nhị phân cho biết giao dịch có phải là gian lận hay không (True: Gian lận, False: Giao dịch hợp pháp)."
    }
    
    # Trả về kết quả nếu tìm thấy, ngược lại thông báo không tồn tại
    print(schema_descriptions.get(col, f"Không tìm thấy thông tin cho cột '{column_name}'."))
```
